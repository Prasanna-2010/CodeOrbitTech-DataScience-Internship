The objective of this task is to clean the Superstore Sales dataset by handling missing values, removing duplicates, correcting data types, and generating summary statistics using the Pandas library.

```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("../datasets/Superstore.csv")
```

```python
df[df["Postal Code"].isnull()]
```

```python
df["Postal Code"]=df["Postal Code"].fillna(5401)
```

```python
df["Postal Code"]=df["Postal Code"].astype(int)
```

```python
df[df["Postal Code"].isnull()]
```

```python
df["Order Date"] =pd.to_datetime(df["Order Date"],dayfirst=True)
df["Ship Date"]=pd.to_datetime(df["Ship Date"],dayfirst=True)
```

## Conclusion

The Superstore dataset was successfully cleaned.

- Missing values in the Postal Code column were handled.
